# Spam text classification

Run from the project directory after installing `requirements.txt`. The bundled data contains SMS messages; email performance is unverified. Duplicate messages are removed before a reproducible, stratified split.


```python
import matplotlib.pyplot as plt
import seaborn as sns
from spam_detector import load_dataset, train_and_evaluate

data = load_dataset()
data.head()
```

```python
data.info()
data["Category"].value_counts()
```

```python
sns.countplot(data=data, x="Category", order=["ham", "spam"])
plt.title("Class distribution after deduplication")
plt.show()
```

## Held-out evaluation

A CountVectorizer/MultinomialNB pipeline learns vocabulary only from training messages. The split uses seed 42 and preserves class proportions.


```python
model, metrics = train_and_evaluate(data)
print(f"Accuracy: {metrics['accuracy']:.4f}")
import pandas as pd
pd.DataFrame(metrics["classification_report"]).T
```

```python
sns.heatmap(metrics["confusion_matrix"], annot=True, fmt="d",
            xticklabels=metrics["label_order"], yticklabels=metrics["label_order"])
plt.title("Held-out confusion matrix")
plt.ylabel("Actual")
plt.xlabel("Predicted")
plt.show()
```
